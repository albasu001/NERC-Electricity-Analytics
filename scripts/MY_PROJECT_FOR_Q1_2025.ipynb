```python
from google.colab import files
import os

uploaded = files.upload()

pdf_file = next(iter(uploaded.keys()))

print("Uploaded file:", pdf_file)
print("File size:", round(os.path.getsize(pdf_file) / (1024 * 1024), 2), "MB")
```

```python
import pdfplumber

pdf = pdfplumber.open(pdf_file)

print("PDF connected successfully.")
print("Number of pages:", len(pdf.pages))
```

```python
all_text = []

for page_number, page in enumerate(pdf.pages, start=1):
    text = page.extract_text()

    if text:
        all_text.append({
            "page": page_number,
            "text": text
        })

print("Pages with extracted text:", len(all_text))
```

```python
search_terms = [
    "Revenue Collection Performance",
    "Meter Deployment by DisCos",
    "Category of complaints received",
    "ATC&C",
    "Revenue Collection",
    "Meter Deployment",
    "complaints"
]

for term in search_terms:
    print("\n" + "=" * 80)
    print("SEARCH:", term)
    print("=" * 80)

    found = False

    for item in all_text:
        if term.lower() in item["text"].lower():
            print("Found on page:", item["page"])
            found = True

    if not found:
        print("Not found")
```

```python
import pandas as pd

extracted_tables = []

for page_number, page in enumerate(pdf.pages, start=1):

    tables = page.extract_tables()

    for table_number, table in enumerate(tables, start=1):

        if table and len(table) > 1:

            df = pd.DataFrame(table[1:], columns=table[0])

            extracted_tables.append({
                "page": page_number,
                "table_number": table_number,
                "data": df
            })

print("Total tables extracted:", len(extracted_tables))
```

```python
for i, item in enumerate(extracted_tables):

    print("\n" + "=" * 100)
    print(f"TABLE INDEX: {i}")
    print(f"PDF PAGE: {item['page']}")
    print(f"TABLE NUMBER: {item['table_number']}")
    print("=" * 100)

    display(item["data"].head(10))
```

```python
for i, item in enumerate(extracted_tables):

    df = item["data"]

    table_text = " ".join(
        df.astype(str).fillna("").values.flatten()
    )

    if "Revenue Collection" in table_text or "Billings" in table_text:

        print("\nFOUND REVENUE TABLE")
        print("Index:", i)
        print("PDF page:", item["page"])

        display(df)
```

```python
for i, item in enumerate(extracted_tables):

    df = item["data"]

    table_text = " ".join(
        df.astype(str).fillna("").values.flatten()
    )

    if "Meter" in table_text or "metered" in table_text.lower():

        print("\nFOUND METERING TABLE")
        print("Index:", i)
        print("PDF page:", item["page"])

        display(df)
```

```python
for i, item in enumerate(extracted_tables):

    df = item["data"]

    table_text = " ".join(
        df.astype(str).fillna("").values.flatten()
    )

    if "complaint" in table_text.lower():

        print("\nFOUND COMPLAINTS TABLE")
        print("Index:", i)
        print("PDF page:", item["page"])

        display(df)
```

```python
for i, item in enumerate(extracted_tables):

    df = item["data"]

    table_text = " ".join(
        df.astype(str).fillna("").values.flatten()
    )

    if "ATC" in table_text.upper() or "energy accounting" in table_text.lower():

        print("\nFOUND ATC&C TABLE")
        print("Index:", i)
        print("PDF page:", item["page"])

        display(df)
```

```python
output_file = "NERC_Q1_2025_RAW_EXTRACTED_TABLES.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:

    for i, item in enumerate(extracted_tables):

        df = item["data"].copy()

        sheet_name = f"Table_{i}_p{item['page']}"

        # Excel sheet names cannot exceed 31 characters
        sheet_name = sheet_name[:31]

        df.to_excel(
            writer,
            sheet_name=sheet_name,
            index=False
        )

print("Saved:", output_file)
```
